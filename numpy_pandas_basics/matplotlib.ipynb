```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
x = np.array([2015, 2018, 2010 , 2026])
y = np.array([10, 5, 20, 40])
plt.plot(x,y)
plt.show()
```

## Customizing plots

```python
line_style = dict(marker="*",
                  markersize=20, 
                  markerfacecolor="#1cd3fc", 
                  markeredgecolor="red", linestyle="dashed", linewidth="2", color="#1c5bfc")
```

```python
y1 = np.array([12, 5, 23, 45])
y2 = np.array([9, 5, 18, 50])


plt.plot(x,y1, **line_style)  ## unpack line_style dict
## . , o v ^ < > 1 2 3 4 8 s p P * h H + x ....
## you can use name, hex kcode, or rgb codes for color
## dashed, dotted, dashdot, none, solid ...
y2 = np.array([17, 23, 20, 30])
plt.plot(x,y2, color="red")
plt.show()
```

## Labels

```python
plt.plot(x,y1)
plt.plot(x,y2)

plt.title("Class size", fontsize=20, family="Arial", fontweight="bold", color="red")
plt.xlabel("year",fontsize=15, color="green")
plt.ylabel("students",fontsize=15, color="green")
plt.xticks(x)
plt.tick_params(axis="both",colors="green")

plt.show()
```

## Grid Lines

```python
x = [1,2,3,4,5]
y = [5,10,15,20,25]
```

```python
plt.grid(axis="x",color="lightyellow",linewidth=2, linestyle="dashed")

plt.plot(x,y)
plt.show()
```

## Bar Charts

```python
categories = np.array(["Grains","Fruit", "Vegies", "Protein", "Sweets"])
values = np.array([4, 3 , 2, 5 ,1])
```

```python
plt.bar(categories,values)
plt.title("daily consumption")
plt.xlabel("food")
plt.ylabel("quantity")
plt.show()
```

```python
plt.barh(categories,values)
plt.show()
```

## Pie Charts

```python
categories = np.array(["Freshmen", "Sophomores", "Juniors", "Seniors"])
values = np.array([300, 250, 175, 222])
color=["red","yellow","blue","green"]
```

```python
plt.pie(values, labels=categories, autopct="%1.1f%%", colors=color, explode=[0,0,0,0.2], shadow=True, startangle=90)
plt.title("College")
plt.show()
```

## Scatter Graphs

```python
x1 = np.array([0,1,1,2,3,4,5,6,7,7,8])
y1 = np.array([55,60,65,62,68,70,75,78,82,85,87])

x2 = np.array([0,1,2,2,3,4,5,6,7,8,8])
y2 = np.array([50,57,65,70,72,72,78,83,88,95,97])
```

```python
plt.scatter(x1,y1, label="Class A")
plt.scatter(x2,y2, label="Class A")
plt.title("Test Scores")
plt.xlabel("Hours Studied")
plt.ylabel("Grade")
plt.legend()
plt.show()
```

## Histograms

```python
scores = np.random.normal(loc=80, scale=10, size=100)    ##loc is median, scale is the standart deviation
scores = np.clip(scores, 0 ,100)
plt.hist(scores, bins=10, edgecolor="black")
plt.show()
```

## Subplots

```python
## Figure =the entire canvas
## Ax = A single plot (subplot)
```

```python
x = np.array([1,2,3,4,5])

figure, axes = plt.subplots(2, 2)
axes[0, 0].plot(x, x*2, color="red")
axes[0,0].set_title("x*2")

axes[0, 1].plot(x, x**2, color="blue")
axes[0,1].set_title("x**2")

axes[0, 1].plot(x, x**2, color="blue")
axes[0,1].set_title("x**2")

axes[1, 0].plot(x, x**3, color="green")
axes[1,0].set_title("x**3")

axes[1, 1].plot(x, x**4, color="purple")
axes[1,1].set_title("x**4")

plt.tight_layout()
plt.show()
```
