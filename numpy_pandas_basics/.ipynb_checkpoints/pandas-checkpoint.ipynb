```python
import pandas as pd
```

```python
##A DataFrame is a 2-dimensional data structure that can store data of different types (including characters, integers, floating point values, categorical data and more) in columns. 
df = pd.DataFrame(
    {
        "Name": [
            "Braund, Mr. Owen Harris",
            "Allen, Mr. William Henry",
            "Bonnell, Miss Elizabeth",
        ],
        "Age": [22, 35, 58],
        "Sex": ["male", "male", "female"],
    }
)
df
##Each column in a DataFrame is a Series
```

## Series

```python
##creating series
series = pd.Series([100,102,104], name="Apartmentes", index=["apartment #1","apartment #2", "apartment #3"])
series
```

```python
print(series.loc)
print(series.loc["apartment #2"])    ##location by label
series.loc["apartment #2"] = 200
print(series.loc["apartment #2"])
```

```python
print(series.iloc[0])    ##integer location
```

```python
##filter
print(series[series >= 103])
```

```python
calories = {"Day 1": 1750, "Day 2":2100, "Day 3":1750}
series2 = pd.Series(calories)
series2
```

```python
series2.loc["Day 3"] += 500
series2
```
