## Creating new columns derived from existing columns

```python
air_quality["london_mg_per_cubic"] = air_quality["station_london"] * 1.882
air_quality.head()
```

```python
air_quality["ratio_paris_antwerp"] = (
    air_quality["station_paris"] / air_quality["station_antwerp"]
)
air_quality.head()
```

```python
## renaming the columns
air_quality_renamed = air_quality.rename(
    columns={
        "station_antwerp": "BETR801",
        "station_paris": "FR04014",
        "station_london": "London Westminster",
    }
)
air_quality_renamed.head()
```

```python
air_quality_renamed = air_quality_renamed.rename(columns=str.lower)

air_quality_renamed.head()
```

### Long to wide table format

```python
air_quality2 = pd.read_csv(
    "data/air_quality_long.csv", index_col="date.utc", parse_dates=True
)

# filter for no2 data only
no2 = air_quality2[air_quality2["parameter"] == "no2"]
# use 2 measurements (head) for each location (groupby)
no2_subset = no2.sort_index().groupby(["location"]).head(2)
no2_subset
```

```python
no2_subset.pivot(columns="location", values="value")   ##The pivot() function is purely reshaping of the data: a single value for each index/column combination is required.
```

```python
##In the case of pivot(), the data is only rearranged. When multiple values need to be aggregated (in this specific case, the values on different time steps), pivot_table() can be used, providing an aggregation function (e.g. mean) on how to combine these values.

air_quality2.pivot_table(
    values="value", index="location", columns="parameter", aggfunc="mean"
)
```

### Wide to long format

```python
no2_pivoted = no2.pivot(columns="location", values="value").reset_index()
no2_pivoted.head()
```

```python
no_2 = no2_pivoted.melt(id_vars="date.utc")
no_2.head()

##The pandas.melt() method on a DataFrame converts the data table from wide format to long format. The column headers become the variable names in a newly created column.
```

## Data Cleaning (fixing/removing incomplete, incorrect or irrelevant data)

```python
titanic = titanic.drop(columns=["Embarked","Parch"])
titanic
```

```python
## handle missing data. if is n/a drop the whole row
titanic = titanic.dropna(subset=["Age"])
titanic
```

```python
titanic = titanic.fillna({"Cabin": "0"})
titanic
```

```python
##fix inconsistent values
titanic["Sex"] = titanic["Sex"].replace({"female": "FEMALE", "male": "MALE"})
titanic
```

```python
## standardize text
titanic["Name"] = titanic["Name"].str.lower()
titanic
```

```python
## fix data types
titanic["Survived"] = titanic["Survived"].astype(bool)
titanic
```

```python
## remove duplicate values
titanic = titanic.drop_duplicates()
titanic
```
