```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.array([2015, 2018, 2010 , 2026])
y = np.array([10, 5, 20, 40])
plt.plot(x,y)
plt.show()
```

## Customizing plots

```python
plt.plot(x,y,marker="*", markersize=20, markerfacecolor="#1cd3fc", markeredgecolor="red", linestyle="dashed", linewidth="2", color="#1c5bfc")
plt.show()
## . , o v ^ < > 1 2 3 4 8 s p P * h H + x ....
## you can use name, hex kcode, or rgb codes for color
## dashed, dotted, dashdot, none, solid ...
```
